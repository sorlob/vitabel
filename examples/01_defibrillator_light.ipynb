# Defibrillator data from out-of-hospital cardiac arrest case

```python
from vitabel import Vitals, Label
case = Vitals()
case.load_data("data/case_1.json")
```

## Plotting Data interactively

```python
plot = case.plot_interactive(
    channels = [
        ["cpr_acceleration"], 
        ["capnography"], 
        ["ecg_pads"], 
        []
    ],
    labels = [
        ["ROSC", "cc_periods"],
        ["etco2_from_capnography", "ROSC"],
        ["ROSC"],
        ["ROSC", "rosc_probability"],
    ],
    channel_overviews = [["cpr_acceleration"]],
    time_unit = "s",
    subplots_kwargs = {"figsize": (16.5, 8)},
)
plot
```
